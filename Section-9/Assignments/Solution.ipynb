```python
def get_no_of_words(file_path):
  with open(file_path, 'r') as file:
    
    words = sum(len(line.split()) for line in file)
    return (words)
docs_len = get_no_of_words('documents.txt')
print(docs_len)
```

```python
def replace_occurences(file_path, word1, word2):
  with open(file_path, 'r') as file:
    data = file.read()
  new_data = data.replace(word1, word2)
  with open(file_path, 'w') as file:
    file.write(new_data)

replace_occurences('data.txt', 'L', 'dream')
```

```python
def read_reverse(file_path):
  with open(file_path, 'r') as file:
    lines = file.readlines()
  for line in reversed(lines):
    print(line.strip())
read_reverse('data.txt')
```

```python
def merge_files(file_list, merged_file):
  for single_file in file_list:
    with open(single_file, 'r') as file:
      content = file.read()
    with open(merged_file, 'a') as merged:
      merged.write(content)

file_list = ['documents.txt', 'destination.txt', 'log.txt']
merge_files(file_list=file_list, merged_file='merge.txt')
```

```python
def split_large_file(file_path, lines_per_file):
  with open(file_path) as file:
    lines = file.readlines()
    for i in range(0, len(lines), lines_per_file):
      with open(f'file_part_{i+1}', 'w') as part_file:
        part_file.writelines(lines[i:i+lines_per_file])
```

```python
def copy_bin(src_file, dest_file):
  with open(src_file, 'rb') as file:
    content = file.read()
  with open(dest_file, 'wb') as file:
    file.write(content)

src_file = 'image.bin'
dest_file = 'copy_image.bin'
copy_bin(src_file, dest_file)
```

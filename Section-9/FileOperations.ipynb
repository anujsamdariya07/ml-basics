```python
# Read a whole file

with open('example.txt', 'r') as file:
  content = file.read()
  print(content)
```

```python
# Read a file line by line

with open('example.txt', 'r') as file:
  for line in file:
    print(line)
```

```python

with open('example.txt', 'r') as file:
  for line in file:
    print(line.strip())# removes the new line character
```

```python
# Writing a file(Overwriting)

with open('example.txt', 'w') as file:
  file.write('Hello World\n')
  file.write('This is a new line.')
```

```python
# Writing a file(Without overwriting)

with open('example.txt', 'a') as file:
  file.write('\nAppend operation taking place.\n')
```

```python
# Writing a list of lines to a file
lines = ['First Line\n', 'Second Line\n', 'Third Line\n']

with open('example.txt', 'w') as file:
  for line in lines:
    file.write(line)
```

```python
# Another way of doing the same
with open('example.txt', 'w') as file:
  file.writelines(lines)
```

```python
# Binary Files

# Writing a binary file
data = b'\x00\x01\x02\x03\x04'
with open('example.bin', 'wb') as file:
  file.write(data)
```

```python
# Reading a binary file
with open('example.bin', 'rb') as file:
  content = file.read()
  print(content)
```

```python
# Read a content from a source file and write it to a destination file
with open('source.txt', 'r') as source_file:
  content = source_file.read()
  with open('destination.txt', 'a') as destination_file:
    destination_file.write(content)
```

```python
# Read a text file and count the no of lines, words and characters
with open('destination.txt', 'r') as file:
  lines = file.readlines()
  line_count = len(lines)
  word_count = sum(len(line.split()) for line in lines)
  char_count = sum(len(line) for line in lines)

  print(f'Line Count: {line_count}')
  print(f'Word Count: {word_count}')
  print(f'Character Count: {char_count}')
```

```python
# WRiting and then reading a file
lines = ['Hello\n', 'Thank You!\n', 'Your welcome\n']
with open('source.txt', 'w+') as file:
  file.writelines(lines)

  # Moving the cursor back to 0th index
  file.seek(0)

  # Print the content of the file
  content = file.read()
  
  print(content)
```
